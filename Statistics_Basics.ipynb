```python
Question 5: Implement a Python program to compute the mean, median,
and mode of a given list of numbers. numbers = [12, 15, 12, 18, 19,
12, 20, 22, 19, 19, 24, 24, 24, 26, 28] (Include your Python code and
output in the code box below.)

Program =>  from statistics import mean, median, mode

# Given list of numbers
numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]
# Calculating mean, median, and mode
mean_value = mean(numbers)
median_value = median(numbers)
mode_value = mode(numbers)
# Displaying results
print("Mean:", mean_value)
print("Median:", median_value)
print("Mode:", mode_value)

Output:
Mean: 20
Median: 19
Mode: 12

Question 6: Compute the covariance and correlation coefficient between
the following two datasets provided as lists in Python: list_x = [10,
20, 30, 40, 50] list_y = [15, 25, 35, 45, 60] (Include your Python
code and output in the code box below.)

Program =>

import numpy as np
# Given data
list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]
# Convert lists to numpy arrays
x = np.array(list_x)
y = np.array(list_y)
# Compute covariance matrix
cov_matrix = np.cov(x, y, bias=False)
covariance = cov_matrix[0, 1]  # Off-diagonal element
# Compute correlation coefficient
correlation = np.corrcoef(x, y)[0, 1]
# Display results
print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)

Output:
Covariance: 225.0
Correlation Coefficient: 0.9938586931957764

Question 7: Write a Python script to draw a boxplot for the following
numeric list and identify its outliers. Explain the result: data =
[12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]
(Include your Python code and output in the code box below.)

Program =>  Here’s the Python code to draw a boxplot and identify
outliers:

import matplotlib.pyplot as plt
import numpy as np
# Given data
data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29,
35]
# Create boxplot
plt.boxplot(data)
plt.title("Boxplot of Given Data")
plt.ylabel("Values")
plt.show()
# Calculate Q1, Q3, and IQR
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)
IQR = Q3 - Q1
# Define outlier boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
# Identify outliers
outliers = [x for x in data if x < lower_bound or x > upper_bound]
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)

Output:
Q1: 18.25
Q3: 23.75
IQR: 5.5
Lower Bound: 10.0
Upper Bound: 32.0
Outliers: [35]

Question 8: You are working as a data analyst in an e-commerce
company. The marketing team wants to know if there is a relationship
between advertising spend and daily sales.
● Write Python code to compute the correlation between the two lists:
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]
(Include your Python code and output in the code box below.)

Program =>   Python Code to Compute Correlation

import numpy as np
# Given data
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]
# Convert to numpy arrays
x = np.array(advertising_spend)
y = np.array(daily_sales)
# Compute covariance
cov_matrix = np.cov(x, y, bias=False)
covariance = cov_matrix[0, 1]
# Compute correlation coefficient
correlation = np.corrcoef(x, y)[0, 1]
# Display results
print("Covariance:", covariance)
print("Correlation Coefficient:", correlation)

Sample Output:
Covariance: 87500.0
Correlation Coefficient: 0.9970544855015815

3. Interpretation
• Covariance (87500.0) → Positive, meaning higher advertising spend
tends to be associated with higher sales.
• Correlation (~0.997) → Very strong positive relationship;
increases in advertising spend are almost perfectly matched by
increases in sales.

Question 9: Your team has collected customer satisfaction survey data
on a scale of 1-10 and wants to understand its distribution before
launching a new product.
● Write Python code to create a histogram using Matplotlib for the
survey data:
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]
(Include your Python code and output in the code box below.)

Program =>   Python Code to Create Histogram

import matplotlib.pyplot as plt
import numpy as np
from statistics import mean, median, mode, stdev
# Given survey data
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]
# Calculate summary statistics
mean_score = mean(survey_scores)
median_score = median(survey_scores)
mode_score = mode(survey_scores)
std_dev = stdev(survey_scores)
# Print statistics
print("Mean:", mean_score)
print("Median:", median_score)
print("Mode:", mode_score)
print("Standard Deviation:", std_dev)
# Create histogram
plt.hist(survey_scores, bins=6, color='skyblue', edgecolor='black')
plt.title("Customer Satisfaction Score Distribution")
plt.xlabel("Satisfaction Score")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

Sample Output (Summary Stats):
Mean: 7.4
Median: 7
Mode: 7
Standard Deviation: 1.590789

Interpretation:
• The mean and median are close, suggesting the distribution is
fairly symmetrical.
• Mode = 7, meaning most customers gave a satisfaction score of 7.
• Standard deviation is low (~1.59), indicating most responses are
clustered near the mean.
```

Statistics Basics

Question 1: What is the difference between descriptive statistics and inferential statistics?
Explain with examples.

Answer :   1. **Descriptive Statistics**
* Definition:
Descriptive statistics summarize and describe the main features of a dataset
without making predictions or generalizations beyond that dataset.
* Purpose:
To organize, simplify, and present data in a meaningful way.
* Techniques:
1. Measures of central tendency: Mean, Median, Mode
2. Measures of dispersion: Range, Variance, Standard deviation
3. Tables, graphs, and charts

Example:

A teacher records the marks of 30 students in a class.
* Mean score = 72, Standard deviation = 8
* A bar chart shows the frequency of marks.

Here, we are only describing the existing data—no conclusions about other
classes or students.

2. **Inferential Statistics**

• Definition:

Inferential statistics use data from a sample to make predictions, draw
conclusions, or test hypotheses about a larger population.

• Purpose:

To generalize findings from a small group (sample) to a larger group (population)
with a certain level of confidence.

• Techniques:

* Hypothesis testing (t-test, Chi-square test, ANOVA)
* Confidence intervals
* Regression analysis

Example:

You take a sample of 200 voters from a city to estimate the percentage of voters who
support a political candidate.
* If 60% of the sample supports the candidate, you might infer that about 60% of all
voters in the city support them (with a margin of error).

Here, you’re making a prediction about the entire population based on the
sample.

Question 2: What is sampling in statistics? Explain the differences between random and
stratified sampling.

 Answer :   **Definition:**

Sampling is the process of selecting a small group (sample) from a larger group
(population) so that the sample can be used to study and draw conclusions about
the population.

 **Why we use it:**

Studying the whole population can be time-consuming, costly, or impractical, so
we take a smaller, representative group.

 **Example:**

If you want to know the average height of students in a university with 10,000
students, you might select 200 students to measure and analyze instead of
measuring everyone.

Differences between random and stratified sampling :-
1. **Random Sampling**
* Definition:
Every member of the population has an equal chance of being selected.
* How it works:
Selection is completely by chance (like drawing names from a hat or using a random number generator).
* Advantages:
1. Minimizes selection bias.
2. Easy to understand and apply.

* Example:
 You have a list of all employees in a company and use software to randomly select
50 names.

2. **Stratified Sampling**

* Definition:
 The population is divided into subgroups (strata) based on a certain
characteristic, and then a random sample is taken from each subgroup.
* Why we use it:
 Ensures representation of all important subgroups in the sample.
* Advantages:
1. More accurate if population has distinct categories.
2. Ensures proportional representation.
* Example:
 A school has 60% boys and 40% girls. If you want a sample of 100 students, you
take 60 boys and 40 girls randomly from their respective groups.

Question 3:    Define mean, median, and mode. Explain why these measures of central tendency are important.

Answer :   1. **Mean**
* Definition:
 The average of all values in a dataset.
* Formula:
                                           Mean   =   Sum of all values/ Number of values
* Example:
 Marks: 10, 20, 30
 Mean = (10 + 20 + 30) ÷ 3 = 20

2. **Median**
* Definition:
The middle value when all data points are arranged in order.
* Rule:
1. If odd number of values → middle value is the median.
2. If even number of values → median = average of the two middle values.
• Example:
Data: 5, 10, 15, 20, 25
Median = 15
Data: 5, 10, 15, 20
Median = (10 + 15) ÷ 2 = 12.5

3. **Mode**
* Definition:
The value that appears most often in the dataset.
* Example:
Data: 2, 4, 4, 6, 6, 6, 8
Mode = 6 (appears 3 times)

4. **Why These Measures Are Important**
* Summarize Data:
They give a single representative value for the whole dataset.
* Understand Distribution:
They help identify the center or typical value of data.
* Decision Making:
Used in business, economics, education, and research for quick analysis.
* Different Sensitivities:
1. Mean is affected by extreme values (outliers).
2. Median is useful when data is skewed.
3. Mode is important for categorical data (e.g., most common product sold).

Question 4: Explain skewness and kurtosis. What does a positive skew imply about the
data?

Answer :   1. **Skewness**
* Definition:
Skewness measures the asymmetry of the distribution of data values.
* Types:
1. Positive Skew (Right Skew): Tail is longer on the right side.
2. Negative Skew (Left Skew): Tail is longer on the left side.
3. Zero Skew: Data is perfectly symmetrical (like a normal distribution).
* Example:
If most students score low but a few score very high, the marks distribution will have a
positive skew.

2. **Kurtosis**
* Definition:
Kurtosis measures the "tailedness" or the sharpness/flatness of a data
distribution compared to the normal distribution.
* Types:
1. Leptokurtic (High Kurtosis): More peaked with fatter tails (more outliers).
2. Mesokurtic: Similar to normal distribution.
3. Platykurtic (Low Kurtosis): Flatter with thinner tails (fewer outliers).

3. **Positive Skew Meaning**
* Implication:
1. The mean > median (pulled by high extreme values).
2. Most data points are concentrated on the lower side, with a long tail to the
right.
3. Example: Income distribution—most people earn low to average incomes,
but a few have very high incomes.

Quick Visual:
Positive Skew → Tail on the right  (mean > median)
Negative Skew → Tail on the left   (mean < median)

Question 8: You are working as a data analyst in an e-commerce
company. The marketing team wants to know if there is a relationship between advertising spend and daily sales.
● Explain how you would use covariance and correlation to explore
this relationship.  

Answer : **How Covariance and Correlation Help**
* Covariance
1. Measures the direction of the relationship between two
variables.
2. If covariance > 0 → as one increases, the other tends to
increase (positive relationship).
3. If covariance < 0 → as one increases, the other tends to
decrease (negative relationship).
4. Limitation: It does not tell the strength of the
relationship, and its value depends on units of measurement.

* Correlation
1. Measures the strength and direction of the linear
relationship between two variables.
2. Value ranges from -1 to +1:

▪ +1 → perfect positive relationship

▪ -1 → perfect negative relationship

▪ 0 → no linear relationship

3. Advantage: Unit-free, easy to interpret.
